```python
myTuple=(20, 20, 2, 30, 4, 4, 4)
x=int(input("Enter a Number:"))
repeat=0

for item in myTuple:
    if item==x:
        repeat+=1

print(x," was found ",repeat," time(s)")
```

```python
myTuple=(20, 20, 2, 30, 4, 4, 4)
x=int(input("Enter a Number:"))
repeat=0

for i in range(len(myTuple)):
    if myTuple[i]==x:
        repeat+=1

print(x," was found ",repeat," time(s)")
```

```python
myTuple=(20, 20, 2, 30, 4, 4, 4)
x=int(input("Enter a Number:"))
found=False

for i in range (len(myTuple)):
    if myTuple[i]==x:
        print(i)
        found=True

if found==False:
    print("not found")
```
