```python
a=int(input("Enter A: "))
b=int(input("Enter B: "))
result=0
while a<=b:
    result+=a
    a+=1
print(result)
```

```python
x=int(input("Enter a Number: "))
count=0
for i in range (1, x+1):
    if x%i==0:
        print(i)
        count=count+1

print("count: ",count)
if count==2:
    print("PRIME")
else:
    print("NOT PRIME")
```
