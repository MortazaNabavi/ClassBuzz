```python
mySum=0
for item in myTuple:
    mySum+=item  #mySum=mySum+item
ave=mySum/len(myTuple)
print(ave)
```

```python
myTuple=(3, 5, 6, 3, 4, 44, 44, 44, 100, 1000)
n=len(myTuple)
theSum=sum(myTuple)
theAverage=theSum/n

differenceTotal=0
for item in myTuple:
    differenceTotal+=(item-theAverage)**2

theVariance=differenceTotal/n

theSTD_Deviation=theVariance**0.5

print(theSTD_Deviation)
```
