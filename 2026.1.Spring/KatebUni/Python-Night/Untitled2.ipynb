```python
n=int(input("Enter N:"))
numbers=set()

for i in range (n):
    numbers.add(int(input(f"#{i+1} Enter a Number: ")))

numbers=sorted(numbers)
print(numbers[-2])
```

```python
physics={"Ali", "Ahmad", "Hadi", "Sobhan", "Baqir"}
math={"Ali", "Sobhan", "Nasir", "Hamed", "Reza"}

print("Physics Class:\n", physics)
print("Math Class:\n", math)
print("Both:\n", math & physics)
print("Only in Physics:\n", physics-math)
print("Only in Math:\n", math-physics)
print("Only in One Class:\n", physics^math)
```

```python
def Maqsoom(n):
    for i in range(1,n+1):
        if (n%i==0):
            print(i, end=" ")

            
Maqsoom(10)
print("\n")
Maqsoom(20)
print("\n")
Maqsoom(30)
```
