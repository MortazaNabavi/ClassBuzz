```python
a=float(input("Enter 1st Number: "))
b=float(input("Enter 2nd Number: "))
c=float(input("Enter 3rd Number: "))
d=float(input("Enter 4th Number: "))

sum=a+b+c+d
average=sum/4

print("مجموع: ",sum)
print(f"Sum is {sum}")
print("اوسط: ",average)
```

```python
Score1=float(input("نمره مضمون اول را وارد کنید: "))
Credit1=float(input("تعداد کریدت مضمون اول را وارد کنید: "))

Score2=float(input("نمره مضمون دوم را وارد کنید: "))
Credit2=float(input("تعداد کریدت مضمون دوم را وارد کنید: "))

Score3=float(input("نمره مضمون سوم را وارد کنید: "))
Credit3=float(input("تعداد کریدت مضمون سوم را وارد کنید: "))

Total_Score=Score1*Credit1+Score2*Credit2+Score3*Credit3
Total_Credit=Credit1+Credit2+Credit3

Average=Total_Score/Total_Credit

print("اوسط نمرات برابر است با ",Average)
```
