```python
qMilk=int(input("Enter QTY for Milk: "))
qRice=int(input("Enter QTY for Rice: "))
qSugar=int(input("Enter QTY for Sugar: "))

fMilk=50
fSugar=70
fRice=160

tMilk=qMilk*fMilk
tRice=qRice*fRice
tSugar=qSugar*fSugar

total=tMilk+tSugar+tRice

discountRate=15
discount=total*discountRate/100

taxRate=20
tax=(total-discount)*taxRate/100

pay=total-discount+tax
print(pay)
```

```python
print("Total Price for Milk: ",tMilk)
print("Total Price for Sugar: ",tSugar)
print("Total Price for Rice: ",tRice)
print("TOTAL: ",total)
print("Discount: ",discount)
print("TAX: ",tax)
print("Payable: ",pay)
```
