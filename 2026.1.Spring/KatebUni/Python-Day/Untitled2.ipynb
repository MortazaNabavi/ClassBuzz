```python
num1=float(input("Enter first Number: "))
opt=input("Enter operator: ")
num2=float(input("Enter second Number: "))

if opt=="*":
    print(num1*num2)
elif opt=="-":
    print(num1-num2)
elif opt=="+":
    print(num1+num2)
elif opt=="/":
    print(num1/num2)
elif opt=="%":
    print(num1%num2)
elif opt=="^":
    print(num1**num2)
else:
    print("Error!")
```

```python
month=int(input("Enter Month as Number: "))

match month:
    case 1|2|3:
        print("Spring")
    case 4|5|6:
        print("Summer")
    case 7|8|9:
        print("Fall")
    case 10|11|12:
        print("Winter")
    case _:
        print("Error")
```
