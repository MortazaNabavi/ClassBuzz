```python
numbers=(20, 30, 40, 50)
```

```python
ave=sum(numbers)/len(numbers)
print(ave)
```

```python
sigma=0
for item in numbers:
    sigma=sigma+(item-ave)**2
var=sigma/len(numbers)
print(var)
```

```python
var=0
for item in numbers:
    var=var+((item-ave)**2)/len(numbers)
print(var)
```

```python
stdDev=var**0.5
print(stdDev)
```

```python
import math
stdDev=math.sqrt(var)
print(stdDev)
```

```python
theTuple=(2, 2, 3, 4, 44, 55, 555, 5555, 55555, 0, -10)
x=int(input("Enter a Number: "))

isInTuple=False
for i in range (len(theTuple)):
    if theTuple[i]==x:
        print(i)
        isInTuple=True
    
if isInTuple==0:
    print("Not Found in Tuple")
```

```python
myTuple=(2, 4, 5, 6, 6, 5, 3, 10,6, -10,6, -1000)
x=int(input("Enter a Number: "))
counter=0
isInTuple=False
for i in range(len(myTuple)):
    if myTuple[i]==x:
        counter=counter+1
        print(counter, "Index: ",i, sep="\t")
        isInTuple=True
        
if isInTuple==0:
    print("Not Found in Tuple")
```
