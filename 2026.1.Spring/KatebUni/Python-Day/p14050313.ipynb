```python
n=int(input("Enter a Number: "))
counter=0
i=1

while i<=n:
    if n%i==0:
        counter=counter+1
    i+=1

if counter==2:
    print(n, " is Prime.")
else:
    print(n, " is NOT Prime.")
```
