چهار عدد به عنوان ورودی دریافت کنید.
مجموع و اوسط آن ها را نمایش دهید.


```python
num1=int(input("Enter 1st Number: "))
num2=int(input("Enter 2nd Number: "))
num3=int(input("Enter 3rd Number: "))
num4=int(input("Enter 4th Number: "))

sum= num1+ num2+ num3 + num4
average= sum/4
print(f"{num1}+{num2}+{num3}+{num4}={sum}")
print(f"{sum}/4={average}")
print(f"SUM: {sum} AVERAGE: {average}")
```

```python
mark1=float(input("Enter Mark of 1st Course: "))
credit1=float(input("Enter Credit of 1st Course: "))

mark2=float(input("Enter Mark of 2nd Course: "))
credit2=float(input("Enter Credit of 2nd Course: "))

mark3=float(input("Enter Mark of 3rd Course: "))
credit3=float(input("Enter Credit of 3rd Course: "))

sum= mark1*credit1 + mark2*credit2 + mark3*credit3
creditTotal=credit1+credit2+credit3
average=sum/creditTotal

print(average)
```

دو عدد دریافت کنید
عدد بزرگ تر را نمایش دهید

```python
x=float(input("Enter 1st Number (X): "))
y=float(input("Enter 2nd Number (Y): "))

if x>y:
    print(x)
else:
    print(y)
```
