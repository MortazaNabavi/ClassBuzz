```python
import math
print("In order to find root(s) of Quadratic Equation, Enter the Coefficients: ")
a=float(input("Enter A: "))
b=float(input("Enter B: "))
c=float(input("Enter C: "))
#a,b,c=float(input("Enter A: ")),float(input("Enter B: ")),float(input("Enter C: "))

delta=b**2-(4*a*c)
print("Delta: ",delta)
if delta<0:
    print("The Equation Has NO Root!")
elif delta>0:
    x1=(-1*b+math.sqrt(delta))/(2*a)
    x2=(-1*b-math.sqrt(delta))/(2*a)
    print("Root 1: ",x1)
    print("Root 2: ",x2)
else:
    x=(-1*b)/(2*a)
    print("Root: ",x)
```
